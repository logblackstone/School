```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
response = requests.get('https://btech-data-analytics.github.io/bridgerland-technical-college/employees_3.html')
```

```python
soup = BeautifulSoup(response.text, 'html.parser')
```

```python
id = []
name = []
age = []

for employee in soup.find('table').find_all('tr', class_='employee'):
  id.append(employee.find_all('td')[0].text)
  name.append(employee.find_all('td')[1].text)
  age.append(employee.find_all('td')[2].text)
```

```python
df = pd.DataFrame({'id': id, 'name': name, 'age': age})

df['age'] = pd.to_numeric(df['age'])
avg_age = df['age'].mean()
print(avg_age)
```

```python
std_dev = df['age'].std()
print(std_dev)
```
