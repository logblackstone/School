```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = housing_data = pd.read_csv('housing.csv')
```

```python
sns.scatterplot(x='Landsize', y='Price', hue='Type', data=housing_data)
plt.title("Land Size vs Price by Property Type")
plt.show()
```

```python
sns.lineplot(data=housing_data.groupby('YearBuilt')['Price'].mean().dropna())
plt.title("Average Price Per Year Built")
plt.show()
```

```python
sns.barplot(x='Regionname', y='Price', data=housing_data, estimator=np.mean)
plt.title("Average Price by Region")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.histplot(housing_data['Price'])
plt.title("Histogram of Property Prices")
plt.show()
```

```python
sns.boxplot(x='YearBuilt', y='Regionname', data=housing_data)
plt.title("Box Plot of Years Built by Region")
plt.show()
```
